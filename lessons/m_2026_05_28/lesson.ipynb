### Пример применения строкового форматирования

```python
# Создание шаблона форматирования  строки
user_info_template = "Имя - {}, возраст - {}, обращайтесь !"

# Данные из внешнего источника (БД, API, локальных фалов)
data_from_db = ["Bob", 22]

# Форматирование строки (приведение строки к нужному нам виду: информация нанизывается на шаблон)
user_info_template.format(*data_from_db)
```
