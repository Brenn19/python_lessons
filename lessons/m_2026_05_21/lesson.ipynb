### Форматирование строк

```python
name = "Иван"
surname = "Иванов"
age = 35
user_info = [name, surname, age]
```

```python
"Имя - {}, Фамилия - {}, возраст - {}".format(name, surname, age)
```

```python
"Имя - {}, Фамилия - {}, возраст - {}".format(*user_info)  # *user_info - name, surname, age
```

```python
user_info = [
    name, surname, age,
    name, surname, age,
    name, surname, age,
    name, surname, age,
    name, surname, age,
]


user_info_for_print = """
Имя - {}, Фамилия - {}, возраст - {},
Имя - {}, Фамилия - {}, возраст - {},
Имя - {}, Фамилия - {}, возраст - {},
Имя - {}, Фамилия - {}, возраст - {},
Имя - {}, Фамилия - {}, возраст - {},
""".format(*user_info)  # *user_info - name, surname, age

print(user_info_for_print)
```
